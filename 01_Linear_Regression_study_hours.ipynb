```python
from sklearn.linear_model import LinearRegression

X = [[1], [2], [3], [4], [5], [6]]
y = [3, 4, 5, 6, 7, 8 ]

model = LinearRegression()
model.fit(X,y)

prediction = model.predict([[8]])

print("Predicted score:", prediction[0])
```
